# Galaxy Scaling Relations

This tutorial notebook will walk you through how to plot two scaling relations for galaxies using basic queries in opencosmo. Lets start by importing the packages we will need.

```python
import numpy as np
import matplotlib.pyplot as plt
import opencosmo as oc
```

We will use opencosmo to read in a the properties of 1,000,000 galaxies with a stellar mass between $M_* = [10^{11},10^{12}] M_\odot$. Once we open the file with opencosmo, many useful galaxy properties are at our fingertips with just some simple column calls. As a sanity check, let's open the data, and plot the distribution of galaxy mass for our galaxies.

```python
data = oc.open('filtered_galaxy_catalog.hdf5')
mass = data.select('gal_mass_star').get_data().value
bins = np.logspace(np.log10(mass.min()), np.log10(mass.max()), 15)
plt.hist(mass, bins = bins)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Stellar Mass [$M_\odot$]')
plt.ylabel('Counts')
plt.show()
```
